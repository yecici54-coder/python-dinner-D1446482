```python
# ==========================================
# Python 資料型別實作作業：聚餐費用試算器
# 姓名：葉又慈
# 學號：D1446482

# ==========================================

print("=== 測試一 (原始資料：4人，折價券100元) ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱 (去除頭尾空白並轉小寫)
clean_event_name = event_name.strip().lower()

# 2. 整理餐費並轉換型別 (移除 NT$、逗號、空白，轉成 float)
clean_meal_amount = meal_amount.replace("NT$", "").replace(",", "").strip()
meal_amount_float = float(clean_meal_amount)

# 將人數轉成整數
people_int = int(people)

# 3. 分步計算費用
service_fee = meal_amount_float * service_rate
total_amount = meal_amount_float + service_fee - discount
cost_per_person = total_amount / people_int

# 4. 判斷是否超過預算 (回傳布林值)
is_over_budget = cost_per_person > budget_per_person

# 5. 輸出費用摘要
print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_amount_float:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_int} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}\n")


print("=== 測試二 (人數改為 5 人，折價券維持 100 元) ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "5" # 修改人數
service_rate = 0.10
discount = 100
budget_per_person = 300

clean_event_name = event_name.strip().lower()
meal_amount_float = float(meal_amount.replace("NT$", "").replace(",", "").strip())
people_int = int(people)

service_fee = meal_amount_float * service_rate
total_amount = meal_amount_float + service_fee - discount
cost_per_person = total_amount / people_int
is_over_budget = cost_per_person > budget_per_person

print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_amount_float:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_int} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}\n")


print("=== 測試三 (人數恢復 4 人，折價券改為 220 元) ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 220  # 修改折價券
budget_per_person = 300

clean_event_name = event_name.strip().lower()
meal_amount_float = float(meal_amount.replace("NT$", "").replace(",", "").strip())
people_int = int(people)

service_fee = meal_amount_float * service_rate
total_amount = meal_amount_float + service_fee - discount
cost_per_person = total_amount / people_int
is_over_budget = cost_per_person > budget_per_person

print(f"聚餐名稱：{clean_event_name}")
print(f"原始餐費：{meal_amount_float:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_int} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}")
```
